# Maze Solver
Generate a random maze and compare BFS, DFS, and A* pathfinding.

This notebook visualizes the maze, animates each solver, and compares exploration counts and path lengths.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'src'))

import matplotlib.pyplot as plt
from IPython.display import HTML
from maze_solver import generate_maze, solve, animate_solve
```

```python
maze = generate_maze(15, 15, loops=25, seed=7)
START, GOAL = (1, 1), (maze.shape[0] - 2, maze.shape[1] - 2)

plt.figure(figsize=(5, 5))
plt.imshow(maze, cmap='binary', interpolation='nearest')
plt.scatter([START[1]], [START[0]], c='limegreen', s=60, label='start')
plt.scatter([GOAL[1]], [GOAL[0]], c='red', s=60, label='goal')
plt.legend(loc='upper right')
plt.axis('off')
plt.show()
```

## Algorithms

| Algorithm | Frontier | Shortest path? | Strategy |
|---|---|---|---|
| BFS | FIFO queue | Yes | Explores outward in layers |
| DFS | LIFO stack | No guarantee | Explores deeply, then backtracks |
| A* | Priority queue | Yes | Uses path cost and Manhattan distance |

```python
results = {name: solve(maze, START, GOAL, name) for name in ('bfs', 'dfs', 'astar')}
for name, (explored, path) in results.items():
    print(f'{name.upper():5} explored: {len(explored):4} | path cells: {len(path) if path else "unreachable"}')
```

```python
for name in ('bfs', 'dfs', 'astar'):
    explored, path = results[name]
    animation = animate_solve(maze, explored, path, title=name.upper())
    display(HTML(animation.to_jshtml()))
```

```python
names = {'bfs': 'BFS', 'dfs': 'DFS', 'astar': 'A*'}
explored_counts = [len(results[a][0]) for a in names]
path_lengths = [len(results[a][1]) if results[a][1] else 0 for a in names]

fig, axes = plt.subplots(1, 2, figsize=(9, 3.5))
axes[0].bar(names.values(), explored_counts)
axes[0].set_title('Cells explored (lower = less work)')
axes[1].bar(names.values(), path_lengths)
axes[1].set_title('Path length (lower = better route)')
plt.tight_layout()
plt.show()
```
